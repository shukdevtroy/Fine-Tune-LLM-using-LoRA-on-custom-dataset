```python
import torch
from transformers import AutoModelForSequenceClassification, AutoTokenizer

# Your custom list of texts
text_list = [
    "I love this product!",
    "This is the worst experience I've ever had.",
    "Not what I expected.",
    "Absolutely fantastic service!",
    "Terrible. Would not recommend.",
    "The quality exceeded my expectations.",
    "I’m really disappointed with the customer support.",
    "It does the job, but nothing special.",
    "Highly recommend to anyone looking for quality.",
    "A complete waste of money.",
    "Works well, and delivery was fast.",
    "I regret buying this.",
    "Totally worth the price!",
    "The app keeps crashing. Very frustrating.",
    "One of the best purchases I’ve made this year."
]

# Label mapping for binary classification
id2label = {
    0: "Negative",
    1: "Positive"
}

# Load model and tokenizer from the fine-tuned directory
model_load_path = "/content/drive/MyDrive/Fine tune model/"
tokenizer_load_path = "/content/drive/MyDrive/Fine tune model/"

model = AutoModelForSequenceClassification.from_pretrained(model_load_path)
tokenizer = AutoTokenizer.from_pretrained(tokenizer_load_path)

# Choose device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

print("Model and tokenizer have been loaded.")
print("Predictions with the loaded model:")
print("--------------------------")

# Make predictions
for text in text_list:
    # Tokenize and move input to the same device
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True).to(device)

    # Disable gradient tracking for inference
    with torch.no_grad():
        logits = model(**inputs).logits

    # Get the predicted class
    predicted_class = torch.argmax(logits, dim=1).item()
    label = id2label[predicted_class]

    print(f"{text} --> {label}")
```
